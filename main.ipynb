Svar: 
Att Python är ett dynamiskt typat språk innebär flera praktiska konsekvenser för hur vi skriver och arbetar med Python-Kod. Här är några viktiga punkter: 
1. Det finns ingen deklaration av data typer, vilket betyder att du inte behöver explicit deklarera datatyper för variabler, då Python själv bestämmar det baserat på det tilldelade värdet. T.ex. om man skriver x = 10 så kommer Python att förstå att x ör en heltal. 
2. Ändring av typer under körning: Då Python är dynamiskt, kan man tilldela ett nytt värde ed en annen datatyp till samma variabel. T.ex. man kan först tilldela ett heltal till en variabel sedan kan man tilldela en sträng till samma variabel senare utan problem 
3. Typkotroll sker vid körning: Python gör typkontrollen vid körning iställer för vid kompilering. Detta betyder att vissa typer av typfel inte upptäcks förrän koden körs. Dette fel kan till ex uppstå om man försöker addera en sträng och ett heltal. 
4. Felixibilitet och enkelhet: Dynamisk typning gör Python mycket flexibel och lättanvänd. Du kan snabbt skriva och ändra kod utan att behöve tänka på typdeklarationer, vilket är användbart för snabb utveckling och prototyping. 
5. Risk för typfel: Flexibiliteten med dynamisk typning innebär ocksp en ökad risk för tpfel. Därför är det viktit att skriva tester och eventuellt använda typkontrollverktyg som mypy för att säkerställa att din kod är typkonsistent. 

````python
# skapa variabler
num1 = 5
num2 = 10

# Använda print() fö ratt skriva ut värdet
print("värdet av num1:", num1)
print("värdet av num2:", num2)

#Addera variblerna och skapa en ny variabel ed resultatet 
sum_result = num1 + num2

# Skriv resultatet av additionen
print("Resultatet av additionen:", sum_result)


# Flexibilitet med dynamisk typning
# Demonstration av typändring under körning
x = 10
print(type(x))      # <class `int`>

x = "hello"
print (type(x))        # <class `str`>

# Typkontroll vid körning
def add(a, b):
    return a + b

print(add(5, 10))       # Detta fungerar och skriver ut 15
print(add(5, "hello"))       # Detta kommer att ge ett typfel vid körning


# Flexibilitet och enkelhet
def print_length(Value):
    print("Längden är:", len(Value))

print_length("Hej")         # Fungerar, eftersom stränger har en längd
print_length([1, 2, 3])     # Fungerar, eftersom listor har en längd
print_length({1: 'a', 2: 'b'})  # Fungerar, eftersom dictionaries har en längd
````

Funktionen `randrange(start, end)` i modulen `random` genererar ett slumpat tal mellan _start_ och _end - 1_. `randrange(5, 10)` ger ger alltså tal mellan 5 och 9.

Använd randrange för att simulera **två** tärningskast (6-sidiga) och returnera summan av kasten.

Gör detta genom att fylla i funktionen `throw_dice()`.

````python
from random import randrange

def throw_dice():
    # Simulera första tärningskastet
    dice1 = randrange(1, 7)
    # Simulera andra tärningskastet
    dice2 = randrange(1, 7)
    # Returnera summan av de två tärningskasten
    return dice1 + dice2

# Anropa funktionen och skriv ut resultatet
print(throw_dice())  # Ska skriva ut ett tal mellan 2 och 12 eftersom det är summen av två tärningar med värden mellan 1 och 6
````

Svara genom att kommentera koden

Keywordet in används två gånger, men i olika sammanhang och med olika syften.

Första användningen av in: Används inom en for-loop för att iterera genom en lista.

Exempel: for name in names:
Syfte: Att iterera genom varje element i listan names.

Andra användningen av in: Används inom en if-sats för att kontrollera om en delsträng finns i en sträng.

Exempel: if "to" in name:
Syfte: Att kontrollera om delsträngen "to" finns någonstans i strängen name.

````python
names = ["tobias", "andreas", "tove"]
new_names = []
for name in names:
   if "to" in name:
      new_names.append(name)


# En lista med namn
names = ["tobias", "andreas", "tove"]

# En tom lista för att lagra nya namn som uppfyller ett visst villkor
new_names = []

# För varje namn i listan names
for name in names:  # Här används "in" för att iterera genom varje element i listan names
    # Om delsträngen "to" finns i namnet
    if "to" in name:  # Här används "in" för att kontrollera om delsträngen "to" finns i strängen name
        # Lägg till namnet i listan new_names
        new_names.append(name)
````

### Uppgift 11

Du har följande dictionary där datan beskriver en mängd studenter. Nyckeln är deras personnnummer, valuen är ytterligare en dictionary med data. Du behöver rensa bort alla studenter som
har blivit godkända. 
Hur går du till väga? Skriv kod och kommentera.

Tips: Du behöver loopa.

Svar: För att rensa bort alla studenter som har blivit godkända från dictionaryn student_grades, kan vi använda en loop för att iterera genom vrje student och sedan ta bort de som har passed, satt till true. 

````python
# Skapa dictionary med studentdata 
student_grades = {
    9210114413 : {"grade": 55, "passed": True},
    9010054514 : {"grade": 22, "passed": False},
    8504129392 : {"grade": 60, "passed": True},
    9201134555 : {"grade": 15, "passed": False},
}

# Skapa en lista för att hålla personnummer för godkända studenter
students_to_remove = []

# Iterera genom varje student i dictionaryn
for personnummer, data in student_grades.items():
    # Kontrollera om studenten har blivit godkänd
    if data["passed"]:
        # Lägg till personnummer till listan för att ta bort
        students_to_remove.append(personnummer)

# Ta bort alla godkända studenter från dictionaryn
for personnummer in students_to_remove:
    del student_grades[personnummer]

# Visa den uppdaterade dictionaryn
print(student_grades)
````

### Uppgift 12

Du har en lista som består av dictionaries. Varje dictionary representerar data för en enskild student i ett system.

Du vill skapa en ny lista som endast består av studenter där "Passed" är True, hur gör du? 

**Tips** Utnyttja listor, loops och if-statements. 

**Svara med kod**

````python
students = [
    {'name': 'John', 'age': 18, 'passed': True, 'major': 'computer science', 'GPA': 3.8},
    {'name': 'Jane', 'age': 19, 'passed': False, 'major': 'biology', 'GPA': 3.2},
    {'name': 'Bob', 'age': 20, 'passed': True, 'major': 'engineering', 'GPA': 3.5},
    {'name': 'Alice', 'age': 21, 'passed': False, 'major': 'history', 'GPA': 2.9},
    {'name': 'David', 'age': 22, 'passed': True, 'major': 'business', 'GPA': 3.6}
]

# Skapa en ny lista för studenter som har passerat
passed_students = []

# Iterera genom varje student i listan students
for student in students:
    # Kontrollera om studenten har passerat
    if student['passed']:
        # Lägg till studenten i listan passed_students
        passed_students.append(student)

# Visa den nya listan med studenter som har passerat
print(passed_students)
````
